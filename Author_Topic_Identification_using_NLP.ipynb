## Author/Topic Identification using NLP

## Importing the data

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel(r'dataset.xlsx')
y = df['target'].values
X = df['data'].values
```

## Importing the stopwords

```python
f2 = open("kannadastopwords.txt","r",encoding="utf-8")
kn_stopwords = f2.readlines()
for i in range(len(kn_stopwords)):
    kn_stopwords[i] = kn_stopwords[i].strip()
f2.close()
```

```python
import nltk
nltk.download('punkt_tab')
```

## Removing stop words

```python
from nltk.tokenize import word_tokenize
dat_pro=[]
for index, entry in enumerate(df['data']):
    processed=[]
    for word in word_tokenize(entry):
        if word not in kn_stopwords:
            processed.append(word)
    dat_pro.append(str(processed))
```

## Splitting the dataset into training set and test set

```python
from sklearn.model_selection import train_test_split
df['data1'] = dat_pro
Train_X, Test_X, Train_Y, Test_Y = train_test_split(df['data1'],df['target'], test_size = 0.2, random_state = 42, stratify = y)
```

## Label Encoding the target values

```python
from sklearn.preprocessing import LabelEncoder
Encoder = LabelEncoder()
#print(Encoder.classes_)
Train_Y = Encoder.fit_transform(Train_Y)
Test_Y = Encoder.transform(Test_Y)
```

## Creating sparse matrix


```python
from sklearn.feature_extraction.text import CountVectorizer
count_vect = CountVectorizer()
# fit1=count_vect.fit_transform(df['data1'])
xtrain_dtm = count_vect.fit_transform(Train_X)
xtest_dtm = count_vect.transform(Test_X)
```

## Fitting the model to the dataset

### Naive Bayes


```python
#NAIVE_BAYES
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, confusion_matrix
Naive = MultinomialNB()
Naive.fit(xtrain_dtm, Train_Y)
predictions_NB = Naive.predict(xtest_dtm)
print("The prediction accuracy of Naive Bayes is: ", accuracy_score(Test_Y, predictions_NB)*100,"%")
#print(confusion_matrix(Test_Y,predictions_NB))
```

### K-fold Cross Validation

```python
from sklearn.model_selection import cross_val_score
MSE_CV = cross_val_score(Naive, xtrain_dtm, Train_Y, cv=10, n_jobs = -1)
print(MSE_CV.mean())
```

### Support vector Machine

```python
# Support Vector Machine
from sklearn.svm import SVC
svclassifier = SVC(kernel='rbf')
svclassifier.fit(xtrain_dtm,Train_Y)
y_pred = svclassifier.predict(xtest_dtm)
#print(confusion_matrix(Test_Y,y_pred))
print("The prediction accuracy of SVM is: ",accuracy_score(Test_Y, y_pred)*100,"%")
```
